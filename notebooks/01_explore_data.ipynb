# Exploratory Data Analysis\n\nLoad `matches.parquet` and inspect goals distribution and home-win rate.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet("../data/processed/matches.parquet")
print(f"Dataset: {df.shape[0]} matches, {df.shape[1]} columns")
df.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Goals distribution
total_goals = df["home_goals"] + df["away_goals"]
axes[0].hist(total_goals.dropna(), bins=range(0, 12), edgecolor="black", alpha=0.7)
axes[0].set_xlabel("Total Goals")
axes[0].set_ylabel("Frequency")
axes[0].set_title("Goals per Match Distribution")

# Home win rate by league
win_rate = df.groupby("league")["result"].apply(lambda x: (x == "H").mean())
win_rate.sort_values().plot.barh(ax=axes[1], color="steelblue")
axes[1].set_xlabel("Home Win Rate")
axes[1].set_title("Home Win Rate by League")
axes[1].axvline(x=0.5, color="red", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
